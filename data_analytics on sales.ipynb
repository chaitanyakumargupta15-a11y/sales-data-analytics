```python
import pandas as pd

# 1. Read CSV file

df = pd.read_csv("synthetic_sales_dirty_15000.csv")

# 2. Understand the dataset

print("First 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())

print("\nShape of dataset:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nDataset information:")
df.info()

# 3. Check missing values

print("\nMissing values:")
print(df.isnull().sum())

# 4. Check duplicate rows

print("\nDuplicate rows:")
print(df.duplicated().sum())

# 5. Create a clean copy

df_clean = df.copy()

# 6. Remove duplicate rows

df_clean.drop_duplicates(inplace=True)

# 7. Remove extra spaces from column names

df_clean.columns = df_clean.columns.str.strip()

# 8. Remove extra spaces from text values

for col in df_clean.select_dtypes(include="object").columns:
    df_clean[col] = df_clean[col].str.strip()

# 9. Replace empty strings with missing values

df_clean.replace("", pd.NA, inplace=True)

# 10. Fill missing numerical values with median

for col in df_clean.select_dtypes(include="number").columns:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# 11. Fill missing text values with mode

for col in df_clean.select_dtypes(include="object").columns:
    if df_clean[col].notna().any():
        df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

# 12. Check missing values again

print("\nMissing values after cleaning:")
print(df_clean.isnull().sum())

# 13. Check duplicates again

print("\nDuplicates after cleaning:")
print(df_clean.duplicated().sum())

# 14. Numerical data analysis

print("\nNumerical analysis:")
print(df_clean.describe())

# 15. Categorical data analysis

print("\nCategorical analysis:")
print(df_clean.describe(include="object"))

# 16. Check unique values

print("\nUnique values:")
print(df_clean.nunique())

# 17. Check value counts of categorical columns

for col in df_clean.select_dtypes(include="object").columns:
    print("\n", col)
    print(df_clean[col].value_counts())

# 18. Check correlation

print("\nCorrelation:")
print(df_clean.corr(numeric_only=True))

# 19. Final dataset

print("\nFinal dataset:")
print(df_clean.head())

print("\nFinal shape:")
print(df_clean.shape)

# 20. Save cleaned CSV

df_clean.to_csv("sales_cleaned.csv", index=False)

print("\nCleaning and analysis completed!")
```
